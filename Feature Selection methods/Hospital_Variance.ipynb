```python
import pandas as pd

# Sample hospital dataset
data = {
    'Age': [25, 45, 52, 36, 61],
    'Blood_Pressure': [120, 140, 135, 130, 145],
    'Cholesterol': [180, 220, 210, 200, 230],
    'Gender': ['M', 'F', 'M', 'F', 'M'],
    'Country_Code': ['India', 'India', 'India', 'India', 'India'],
    'Hospital_ID': [101, 101, 101, 101, 101],
    'Machine_Version': ['v2.1', 'v2.1', 'v2.1', 'v2.1', 'v2.2'],
    'Disease': [0, 1, 1, 0, 1]
}

df = pd.DataFrame(data)
```

```python
df_encoded = pd.get_dummies(df, drop_first=True)
```

```python
from sklearn.feature_selection import VarianceThreshold

# Remove features with very low variance
selector = VarianceThreshold(threshold=0.01)

X = df_encoded.drop('Disease', axis=1)
y = df_encoded['Disease']

X_selected = selector.fit_transform(X)
```

```python
selected_features = X.columns[selector.get_support()]
print("Selected Features:\n", selected_features)
```

```python
removed_features = X.columns[~selector.get_support()]
print("Removed Features:\n", removed_features)
```

```python
X_final = pd.DataFrame(X_selected, columns=selected_features)
```
